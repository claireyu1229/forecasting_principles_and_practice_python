```python
a10 = pbs.loc[pbs["ATC2"] == "A10"]
a10
```

```python
pbs_new = (pbs.loc[pbs["ATC2"] == "A10"].drop(columns=["ATC1", "ATC2"]))
```

```python
pbs_new
```

```python
total_cost_df = (
    pbs.loc[pbs["ATC2"] == "A10"]
    .drop(columns=["ATC1", "ATC2"])
    .groupby("Month", as_index=False)
    .agg({"Cost": "sum"})
    .assign(Cost = lambda x:(x["Cost"]/1e6) .round(2))
)
```
